```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Original dataset:", X.shape)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Feature Scaling:
# The features have different numerical ranges.
# For example:
# - Alcohol: approximately 11–15
# - Magnesium: approximately 70–160
# - Proline: approximately 278–1680
#
# These differences in scale can affect the performance of certain machine learning algorithms.
# Feature scaling brings all numerical features to a comparable scale, ensuring that no feature dominates others simply because of its larger numerical values.

# ---

# 1. StandardScaler:

# StandardScaler is used to standardize numerical features by removing the mean and scaling them to unit variance.

# Formula:
# z = (x - mean) / standard_deviation

# After scaling, each feature will approximately have:
# - Mean = 0
# - Standard Deviation = 1
#
# This ensures that all features are on a comparable scale and prevents features with larger numerical ranges from dominating the machine learning algorithm.

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Verify scaling

scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)

scaled_df.describe().round(2)
```

```python
# 2. MinMaxScaler:

# MinMaxScaler is a feature scaling technique that converts feature values into a specified range, usually between 0 and 1.
# It works by subtracting the minimum value from each value and dividing it by the difference between the maximum and minimum values.

# Formula:
# x_scaled = (x - x_min) / (x_max - x_min)

from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()

X_train_minmax = minmax_scaler.fit_transform(X_train)
X_test_minmax = minmax_scaler.transform(X_test)

# Which algorithms need feature scaling?

# Logistic Regression      : Scaling required
# K-Nearest Neighbors      : Scaling required
# Support Vector Machine   : Scaling required
# Naive Bayes              : Usually not required
# Decision Tree            : Not required
# Random Forest            : Not required
# Gradient Boosting        : Usually not required
# Neural Network           : Scaling required

# Note:
# Distance-based and gradient-based algorithms generally
# benefit from feature scaling, while tree-based algorithms
# usually do not need scaling.

# Train Classification Models:

# 1. Logistic Regression: Used as a basic linear classification model.
# 2. K-Nearest Neighbors (KNN): Classifies data based on nearby data points.
# 3. Support Vector Machine (SVM): Finds the best decision boundary to classify data.
# 4. Decision Tree: Makes predictions using rule-based decisions.
# 5. Random Forest: Combines multiple decision trees to improve prediction performance.
```

```python
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000))
    ]),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=5))
    ]),

    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC())
    ]),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    )
}
```

```python
trained_models = {}

for name, model in models.items():
    model.fit(X_train, y_train)

    trained_models[name] = model

    print(f"{name} trained successfully")
```

```python
predictions = {}

for name, model in trained_models.items():
    predictions[name] = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

for name, y_pred in predictions.items():
    accuracy = accuracy_score(y_test, y_pred)

    print(f"{name}: {accuracy:.4f}")
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

results = []

for name, y_pred in predictions.items():

    results.append({
        "Model": name,
        "Accuracy Score": accuracy_score(y_test, y_pred),

        "Accuracy Percentage": round(accuracy_score(y_test, y_pred) * 100, 2),

        "Precision (Macro)": precision_score(
            y_test, y_pred, average="macro"
        ),

        "Recall (Macro)": recall_score(
            y_test, y_pred, average="macro"
        ),

        "F1 (Macro)": f1_score(
            y_test, y_pred, average="macro"
        ),

        "F1 (Weighted)": f1_score(
            y_test, y_pred, average="weighted"
        )
    })

results_df = pd.DataFrame(results)

results_df.round(4)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

for name, y_pred in predictions.items():

    cm = confusion_matrix(y_test, y_pred)

    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=sorted(y.unique())
    )

    display.plot(cmap="Blues")

    plt.title(f"{name} - Confusion Matrix")
    plt.show()
```

```python
from sklearn.metrics import classification_report

for name, y_pred in predictions.items():

    print(f"\n{name}")
    print("=" * 60)

    print(classification_report(
        y_test,
        y_pred,
        zero_division=0
    ))
```

```python
# Extract per-class metrics for comparison
# Instead of manually reading every classification report, create a comparison table.

class_results = []

for name, y_pred in predictions.items():

    report = classification_report(
        y_test,
        y_pred,
        output_dict=True,
        zero_division=0
    )

    for class_label in sorted(y.unique()):

        metrics = report[str(class_label)]

        class_results.append({
            "Model": name,
            "Class": class_label,
            "Precision": metrics["precision"],
            "Recall": metrics["recall"],
            "F1-score": metrics["f1-score"],
            "Support": metrics["support"]
        })

class_results_df = pd.DataFrame(class_results)

class_results_df.round(4)
```

```python
# Compare class performance visually

plt.figure(figsize=(10, 6))

sns.barplot(
    data=class_results_df,
    x="Class",
    y="F1-score",
    hue="Model"
)

plt.title("Per-Class F1-score Comparison")
plt.ylim(0, 1.05)
plt.show()
```
